```python
import pandas as pd
import matplotlib.pyplot as plt



csv_path = "qaoa_vs_lp_copy.csv"

df = pd.read_csv(csv_path)

# Make sure instance matches strings like "01", "02", ...
df["instance"] = df["instance"].astype(str).str.zfill(2)
```

```python
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df_p3 = df[df["p"] == 3].copy()

plt.figure(figsize=(12, 5))

sns.barplot(
    data=df_p3,
    x="instance",
    y="ratio",
    color="steelblue"
)

plt.axhline(y=1, color="red", linestyle="--", label="Optimal")

plt.xlabel("Instance")
plt.ylabel("QAOA / Optimal")
plt.title("QAOA Approximation Ratio by Instance (p=3)")
plt.ylim(0, 1.05)
plt.legend()
plt.tight_layout()
plt.show()
```
