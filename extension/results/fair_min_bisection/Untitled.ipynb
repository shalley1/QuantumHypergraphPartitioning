```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("aon_standard_comparison.xlsx")

# Sort by dataset and instance
df = df.sort_values(["dataset", "instance"]).reset_index(drop=True)

x = np.arange(len(df))



plt.figure(figsize=(12, 5))

plt.plot(x, df["optimal_t"],
         label="LP Optimal", color="black",
         linestyle="--", linewidth=2)

plt.plot(x, df["fair_cut_AoN"],
         label="Fair Cut AoN", marker="o", markersize=4)

plt.plot(x, df["fair_cut_standard"],
         label="Fair Cut Standard", marker="s", markersize=4)

plt.xlabel("Instance")
plt.ylabel("Objective Value")
plt.title("Fair Min-Bisection: QAOA vs LP Optimal")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


```

```python

plt.figure(figsize=(12, 5))

plt.axhline(y=1, color="black", linestyle="--",
            linewidth=2, label="Optimal (1.0)")

plt.plot(x, df["ratio_AoN"],
         label="AoN QAOA", marker="o", markersize=4)

plt.plot(x, df["ratio_standard"],
         label="Standard QAOA", marker="s", markersize=4)

plt.xlabel("Instance")
plt.ylabel("Approximation Ratio")
plt.title("Fair Min-Bisection: Approximation Ratios")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
