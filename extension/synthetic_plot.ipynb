```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Approximation ratios (%)
ratios = np.array([
    [99.9, 100.0, 100.0],   # Max Bisection
    [99.7, 99.8, 99.9],   # Min Bisection
    [98.9, 99.6, 99.9],   # Fair Max Bisection
    [85.3, 90.7, 100.0]  # Fair Min Bisection
])

problems = [
    "Max Bisection",
    "Min Bisection",
    "Fair Max Bisection",
    "Fair Min Bisection"
]

hyperedge_sizes = [r"$\mu=3$", r"$\mu=4$", r"$\mu=5$"]

fig, ax = plt.subplots(figsize=(8, 5))

sns.heatmap(
    ratios,
    annot=True,
    fmt=".1f",
    cmap="viridis",
    vmin=90,
    vmax=100,
    xticklabels=hyperedge_sizes,
    yticklabels=problems,
    cbar_kws={"label": "Approximation Ratio (%)"},
    linewidths=0.5,
    linecolor="white",
    ax=ax
)

ax.set_xlabel("Expected Hyperedge Size", fontsize=12)
ax.set_ylabel("Optimization Problem", fontsize=12)
ax.set_title("AoN QAOA Performance on Poisson Hypergraphs",
             fontsize=13, pad=15)

plt.tight_layout()
plt.savefig("poisson_aon_heatmap.pdf", bbox_inches="tight")
plt.savefig("poisson_aon_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```
