```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "/Users/stevehalley/Code/GitHub/QuantumHypergraphPartitioning/extension/results/max_bisection/congress_qaoa_max_bisection_run2.csv"

df = pd.read_csv(file_path)

# Sort by instance
df = df.sort_values("instance")

fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(
    df["instance"],
    df["ratio"],
    marker="o",
    linestyle="-",
    linewidth=1.5,
    markersize=5,
    label="QAOA"
)

# Optimal approximation ratio
ax.axhline(
    y=1.0,
    color="red",
    linestyle="--",
    alpha=0.7,
    label="Optimal"
)

ax.set_xlabel("Instance")
ax.set_ylabel("Ratio to Optimal")
ax.set_title("QAOA Max Bisection — Congress")
ax.grid(True, alpha=0.3)
ax.legend()

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
