```python
def max_cut_from_Q(Q, 
                   seed: int,
                   num_layers: int,
                   scaling_constant: float,
                   step_size: float, 
                   max_trials: int,
                   threshold: float,
                   #non_improving_steps: int
                  ):
    rng = np.random.default_rng(seed=seed)
    n_wires = Q.shape[0] + 1
    triu_indices = np.triu_indices(Q.shape[0])
    edge_list = []
    for (u, v) in zip(triu_indices[0], triu_indices[1]):
        if Q[u][v] == 0.0:
            continue
        else:
            if u == v:
                edge_list.append((0, n_wires - 1))
            else:
                edge_list.append((u, v))
    beta_params = scaling_constant * rng.random([1, num_layers], requires_grad=True)
    gamma_params = scaling_constant * rng.random([1, num_layers], requires_grad=True)
    print(beta_params)
    def U_B(beta):
        for wire in range(n_wires):
            qml.RX(2 * beta[0], wires=wire)
    def U_C(gamma):
        for u, v in edge_list: 
            w  = Q[u][u] if v == (n_wires - 1) else Q[u][v]
            qml.CNOT(wires=(u, v))
            qml.RZ(-w * gamma[0], wires=v)
            qml.CNOT(wires=(u, v))
    dev = qml.device('lightning.qubit', wires=n_wires)
    @qml.qnode(dev)
    def circuit(gammas, betas):
        for wire in range(n_wires):
            qml.Hadamard(wires=wire)
        for k in range(num_layers):
            U_C(gammas[:, k])
            U_B(betas[:, k])
        return [qml.expval(qml.PauliZ(u) @ qml.PauliZ(v)) for u, v in edge_list]
    def objective(gammas, betas):
        expectation_values = circuit(gammas, betas)
        expectation_values = np.array(expectation_values)
        expectation_values = (1/2)*(1-expectation_values)
        return -np.sum(expectation_values)

    opt = qml.AdamOptimizer(stepsize=step_size)
    objective_list = [objective(gamma_params, beta_params)]
    params_list = [(gamma_params, beta_params)]
    for i in range(max_trials):
        (gamma_params, beta_params), current_obj = opt.step_and_cost(objective, gamma_params, beta_params)
        if (i + 1) % 5 == 0:
            print("Current obj: ", current_obj)
            if np.abs(current_obj - objective_list[-1]) <= threshold:
                objective_list.append(current_obj)
                params_list.append((gamma_params, beta_params))
                break
        objective_list.append(current_obj)
        params_list.append((gamma_params, beta_params))
    result = [objective_list, params_list]
    best_params.append(result[1][np.argmin(result[0])])
    best_qaoa_objective = np.min(result[0])
    return best_params, best_qaoa_objective, result
```

```python
step_size = 1e-2
threshold = 1e-8
max_trials = 500
scaling_constant = 5e-2
```

```python
best_params, best_qaoa_objective, result  =  max_cut_from_Q(Q = V, 
                                                            seed = 42,
                                                            num_layers = 2,
                                                            scaling_constant = scaling_constant,
                                                            step_size = step_size, 
                                                            max_trials = max_trials,
                                                            threshold = threshold)
```
