```python
import itertools
import numpy as np
```

```python
def aon_cost(x, edge):
    vals = [x[i] for i in edge]
    #return 1 if edge cut else 0
    return int(max(vals) != min(vals))
```

```python
def total_cost(x, edges):
    return sum(aon_cost(x, e) for e in edges)
```

```python
edges = [
    (0, 1, 2),
    (3, 4, 5),
    (0, 3)    
]
```

```python
def balanced_states(n):
    states = []

    for ones in itertools.combinations(range(n), n // 2):
        x = np.zeros(n, dtype=int)
        x[list(ones)] = 1
        states.append(x)

    return states
```

```python
n = 6

results = []

for x in balanced_states(n):
    cost = total_cost(x, edges)
    results.append((x.copy(), cost))

results.sort(key=lambda pair: pair[1])

for x, cost in results:
    print("".join(map(str, x)), cost)
```

```python
best_cost = results[0][1]

optimal_states = [
    x for x, cost in results
    if cost == best_cost
]

print("Best cost:", best_cost)

for x in optimal_states:
    print(x)
```
